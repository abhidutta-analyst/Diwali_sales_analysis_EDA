# Diwali Sales Data Analysis
This notebook performs exploratory data analysis on the Diwali Sales dataset to understand customer behavior and sales patterns.

```python
# Import necessary libraries for data analysis and visualization
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns

# Set theme for better visualization aesthetics
sns.set_theme(style="whitegrid", palette="pastel")
%matplotlib inline
```

```python
# Load the dataset. Using unicode_escape to handle special characters in the CSV
df = pd.read_csv('Diwali Sales Data.csv', encoding= 'unicode_escape')
```

### Data Cleaning and Preprocessing

```python
# Drop columns that are entirely empty or unrelated to analysis
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)

# Drop rows with null values
df.dropna(inplace=True)

# Change 'Amount' column data type to integer for better numerical analysis
df['Amount'] = df['Amount'].astype('int')
```

### Exploratory Data Analysis

```python
# 1. Gender Analysis
plt.figure(figsize=(6,5))
ax = sns.countplot(x='Gender', data=df)
plt.title('Count of Buyers by Gender')
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()

plt.figure(figsize=(6,5))
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender', y='Amount', data=sales_gen)
plt.title('Total Amount Spent by Gender')
plt.show()
```

**Insight:** Most buyers are female, and their total spending is also significantly higher than males.

```python
# 2. Age Group Analysis
plt.figure(figsize=(8,5))
ax = sns.countplot(data=df, x='Age Group', hue='Gender')
plt.title('Count of Buyers by Age Group and Gender')
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()

plt.figure(figsize=(8,5))
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y='Amount', data=sales_age)
plt.title('Total Amount Spent by Age Group')
plt.show()
```

**Insight:** The 26-35 age group, particularly women, constitutes the largest segment of buyers and contributes most to total sales.

```python
# 3. State-wise Analysis
plt.figure(figsize=(15,6))
sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.barplot(data=sales_state, x='State', y='Orders')
plt.title('Top 10 States by Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

**Insight:** Uttar Pradesh, Maharashtra, and Karnataka are the top-performing states in terms of orders.
